```python
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from PIL import Image

# ===============================
# STEP 1: Upload Image
# ===============================
uploaded = files.upload()
image_name = list(uploaded.keys())[0]

# ===============================
# STEP 2: Convert Image to Grayscale
# ===============================
img = Image.open(image_name).convert("L")

# Convert image to numpy matrix
img_matrix = np.array(img)

print("Original Image Shape:", img_matrix.shape)

# ===============================
# STEP 3: Define 2x2 Convolution Kernel
# ===============================
# This is a simple averaging kernel
kernel = np.array([[1, 1],
                   [1, 1]])

# Normalize kernel (optional but recommended)
kernel = kernel / 4

print("Kernel:\n", kernel)

# ===============================
# STEP 4: Perform Convolution (No Padding)
# ===============================
h, w = img_matrix.shape

# We reduce size by 1 because 2x2 kernel
output_height = h - 1
output_width = w - 1

conv_output = np.zeros((output_height, output_width))

# Sliding window
for i in range(output_height):
    for j in range(output_width):

        # Extract 2x2 region
        region = img_matrix[i:i+2, j:j+2]

        # Element-wise multiplication
        conv_value = np.sum(region * kernel)

        # Store result
        conv_output[i, j] = conv_value

print("Convolution Output Shape:", conv_output.shape)

# ===============================
# STEP 5: Apply Activation Function
# ===============================
# Example: ReLU (Rectified Linear Unit)
# ReLU(x) = max(0, x)

relu_output = np.maximum(0, conv_output)

# ===============================
# STEP 6: Reduce to ONE Final Number
# ===============================
# Example: Take mean of entire convolved matrix

final_number = np.mean(relu_output)

print("Final Single Output Number:", final_number)
```
