```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_excel("/content/india to qatar regional data.xlsx")

# Qatar policies
policies = {
    "World Cup Infrastructure Program (2011)": 2011,
    "Kafala Reform (2018)": 2018,
    "Exit Permit Removal (2020)": 2020,
    "Labor Mobility Reform (2020)": 2020,
    "Minimum Wage Law (2021)": 2021
}

# State columns
state_cols = [c for c in df.columns if c != "Year"]

for policy, year in policies.items():

    # Data after policy implementation
    after = df[df["Year"] >= year]

    # Average migration of each state
    avg_state = after[state_cols].mean().sort_values(ascending=False)

    # Top 10 states
    top10 = avg_state.head(10)

    print("\n")
    print("="*60)
    print(policy)
    print("="*60)
    print(top10)

    # Plot
    plt.figure(figsize=(10,6))

    plt.bar(
        top10.index,
        top10.values,
        width=0.45,
        color="royalblue",
        edgecolor="black"
    )

    plt.title(f"Top 10 Indian States Migrating to Qatar\nAfter {policy}",
              fontsize=14,
              fontweight='bold')

    plt.xlabel("State", fontsize=12)
    plt.ylabel("Average Migration", fontsize=12)

    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.4)

    plt.tight_layout()
    plt.show()
```
